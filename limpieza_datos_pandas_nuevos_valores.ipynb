# 1. Carga del Dataset e Inspección Inicial

En este paso cargamos el archivo CSV `datos_nuevos_practica_pandas.csv` en un DataFrame de Pandas y realizamos una primera inspección estructural para detectar:
- La cantidad total de filas y columnas.
- Los tipos de datos de cada columna.
- La presencia de valores nulos (faltantes).

```python
import pandas as pd

# 1. Cargar el archivo CSV en un DataFrame
df = pd.read_csv('datos_nuevos_practica_pandas.csv')

# 2. Ver las primeras 5 filas para comprobar que cargó bien
print("---Primeras filas---")
print(df.head())
print("\n" + "="*50 + "\n")

# 3. Inspección estructural
# qué columnas tienen valores nulos (NaN) y el tipo de dato
print("---Información del DataFrame---")
print(df.info())
print("\n" + "="*50 + "\n")

```

# 2. Detección de Valores Nulos (Faltantes)

Gracias al método `.info()`, pudimos detectar de forma automática dónde están los baches en nuestros datos:
- **Fecha:** Tiene 31 datos válidos (1 valor nulo/vacío).
- **Calorías:** Tiene 30 datos válidos (2 valores nulos/vacíos).

En los siguientes pasos decidiremos cómo limpiar o rellenar estos valores faltantes.

```python
# 1. Verificamos rápidamente los valores nulos con un comando extra de Pandas
# Esto nos devuelve un 'True' (si es nulo) o 'False' (si tiene datos) sumado por
# columna
print("---Valores nulos por columna---")
print(df.isnull().sum())
print("\n" + "="*50 + "\n")

#Al no ser tan grande el DataFrame vemos sus datos
print("--- VISTA PREVIA DEL DATAFRAME ---")
print(df.to_string())


#(buscamos filas con duración mayor a 120 minutos, por ejemplo)
print("--- FILAS CON DURACIÓN ANÓMALA (> 120 min) ---")
print(df[df['Duración'] > 120])

print("\n" + "="*50 + "\n")

# 2. Corregimos el valor atípico de forma manual (reemplazamos el 450 por 45)
# Usamos .loc para ubicar exactamente la fila 7 y la columna 'Duración'
df.loc[7, 'Duración'] = 45

# 3. Comprobamos que el cambio se haya aplicado bien
print("--- VERIFICACIÓN DE LA FILA 7 CORREGIDA ---")
print(df.loc[7])


```

# 3. Limpieza y Relleno de Valores Nulos

Aplicamos estrategias del mundo real para solucionar los datos faltantes sin perder registros:
- **Fecha:** Rellenamos el valor nulo copiando el valor de la fila anterior (método forward fill), manteniendo la continuidad temporal.
- **Calorías:** Rellenamos los valores nulos calculando el promedio (media) de toda la columna.

```python


# 1. Rellenamos con nulo de 'Fecha' copiando el valor de arriba(forma moderna)

df['Fecha'] = df['Fecha'].ffill()

# 2. Convertimos a fecha real (pasa de object a datatime)
df['Fecha'] = pd.to_datetime(df['Fecha'], format='mixed')

#---------------------------------------------------------------------
# Otra opcion Transformamos la columna 'Fecha' a formato datetime real
# Usamos format='mixed' por si hay pequeñas variaciones en cómo están escritas
#las fechas es otra opcion
#df['Fecha'] = pd.to_datetime(df['Fecha'], format='mixed')

# Eliminamos las filas donde la columna 'Fecha' tenga un valor NaT (vacío)
#df.dropna(subset=['Fecha'], inplace=True)
#---------------------------------------------------------------------

# 3. Rellenar los nulos de Calorías usando el promedio de la columna
promedio_calorias = df['Calorías'].mean()
df['Calorías'] = df['Calorías'].fillna(promedio_calorias)


# 4. Comprobación final: Verificamos que ya no queden nulos
print("--- CANTIDAD DE NULOS DESPUÉS DE LIMPIAR ---")
print(df.isnull().sum())

print("\n" + "="*50 + "\n")

# 5. Comprobamos los tipos de datos actualizados
print("--- TIPOS DE DATOS ACTUALIZADOS (df.info) ---")
df.info()



# Mostramos el resultado para ver cómo quedó
#print("--- VISTA PREVIA DEL DATAFRAME LIMPIO ---")
#print(df.to_string())
```

# 5. Detección y Eliminación de Duplicados

A veces los datasets pueden contener filas repetidas. Antes de hacer análisis estadísticos o gráficos, es fundamental buscarlas y eliminarlas para no distorsionar los resultados.

```python
# 1. Buscamos si hay filas duplicadas (duplicated() nos devuelve True si está repetida)
print("--- CANTIDAD DE FILAS DUPLICADAS ENCONTRADAS ---")
print(df.duplicated().sum())

# 2. Nos aparece una fila, la eliminamos con drop_duplicates()
# Usamos inplace=True para que aplique el cambio directamente en nuestro DataFrame
df.drop_duplicates(inplace=True)

# 3. Comprobamos el tamaño final de nuestro DataFrame ya limpio
print("\n--- TAMAÑO FINAL DEL DATAFRAME LIMPIO ---")
print(df.shape) # Muestra (filas, columnas)
```
